#### 3.Indexing in Series:
#### i. What is index?
- Index means, the position of value where it shows.
- The index is the core feature in pandas.
- By default,index is assigned to enery value.
- From the o/p ,the  left most column is the index column.
- The generic name for an index is an axis.
#### ii. Tndex default values:
- The default values for an index  are integers.
- The index starts from 0,1,2,3 etc.
- Based on requirement we can customize the index.
- These are called as axis label.

```python
#customise index:
import pandas as pd
a=[1,2,3,4]
n=[122,154,784,3001]
s=pd.Series(a,index=n)
print(s)
```

```python
import pandas as pd
a=[1,2,3,4]

s=pd.Series(a,name='Count')
print(s)
```

```python
# giving index and name:
import pandas as pd
a=[1,2,3,4]
n=[122,154,784,3001]
s=pd.Series(a,name="lion",index=n)
print(s)
```

#### here, name and index are the keyword arguments.

```python
# giving name as index
import pandas as pd
price=[111,200,333,400]
n=["poil","koil","poty","rotty"]
s=pd.Series(price,index=n)
print(s)
```

#### 4.Accessing values in series:
- we can access series values by using index.

```python
import pandas as pd
price=[111,200,333,400]

s=pd.Series(price,name="marks")
print(s)
print()
print(s[0])
print(s[1])
```

```python
# creating Series accessing values
import pandas as pd
p=[1,2,3,4]
k=["kilo","urin","pery","dory"]
s=pd.Series(p,name="marks",index="k")
print(s)
print(s["kilo"])
```
