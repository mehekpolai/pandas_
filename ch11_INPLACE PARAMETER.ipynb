#### 1.The inplace parameter:
- Inplace is a parameter  for dataframe methods like,
  * rename(p,inplace)
  * drop(p,inplace)
  * sort_values(p,inplace)
  * set_index(p,inplace)
- This inplace parameter accepts two  values.
  * inplace=False means it can't overwite the existing dataframe.
  * inplace=True means it can overwite the existing dataframe.
#### simple word:
- inplace means that you should update the orginal object instead of creating new object.
#### 2.rename(p) method- without inplace parameter:

- We can access rename() by using dataframe object.
- We can even change multiple names by using new object.

#### 3.rename(p) method -with inplace parameter:
- rename(p,inplace=True) is a predifined method in dataframe.
- We can provide inplace parameter to rename(p,inplace=True) method.
- If we provide inplace parameter then directly changes wil be apply on orginal dataframe.
- Whenever if we apply inplace=True,then it should update the orginal dataframe instead of creating new dataframe.

```python
# not creating new object:
import pandas as pd
file = r"C:\Users\mehek\OneDrive\Desktop\DF File\sale4.csv"

df=pd.read_csv(file)
print(df.head())
d = {
    "emp id": "Employee_id",
    "emp name": "Employee_name",
    "dept name": "Department Name",
    "sal": "Salary",
    "age": "Age"
}
df.rename(columns=d,inplace=True)
print(df.head())
```
