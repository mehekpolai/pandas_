#### 1. Date data type:
→ Whenever contains we load csv file, if that file contain any column having date values, then by default pandas will consider that column as object.

→ We can provide parse_dates = ["name of the column1", "name of the column2" ] then pandas considered those columns as datetime data type.

```python
# Load file
import pandas as pd

df = pd.read_csv("cust.csv")

print(df)
print(df.dtypes)
```

- By default, pandas will assign "Purchase_Date" column datatype as object (string) date type, mean we can apply only string operations. so we have to convert to datetime data type.

```python
# Loading csv file using "parse_dates"  parameter
import pandas as pd

df = pd.read_csv("cust.csv", parse_dates=["Purchase_Date"])

print(df)
print(df.dtypes)
```

* Here, parse_dates ---> Keyword argument

```python
# Loading csv file using "parse_dates"  & give 2 values to this parameter
import pandas as pd
f=["Purchase_Date","Product_Cost"]
df = pd.read_csv("cust.csv", parse_dates=f)

print(df)
print(df.dtypes)
```

#### 2. Converting object data types into date data type:

- We can convert from object datatype into datetime data type explicity.
- By using ,
   * to_datetime(p) function
   * astypes(p) method 
#### i. to_datetime(p) function:
→ to datetime(p) is predefined function in pandas.

→ This function we should access by using pandas library.

→ This function convert from object data type into date data type.

```python
# Loading csv file using "Purchase_Date" function
import pandas as pd

df = pd.read_csv("cust.csv")
df.info
df["Purchase_Date"]=pd.to_datetime(df["Purchase_Date"])
print(df)
print(df.dtypes)
```

#### ii.astypes(p) method :
→ It is predefined method Series class.

→ This method we should access by using Series object only.

→ This method convert from object data type into datetime datatype.

```python
# Loading csv file using "astypes(p)" method
import pandas as pd

df = pd.read_csv("cust.csv")
df.info
df["Purchase_Date"]=(df["Purchase_Date"].astype("datetime64[ns]"))
print(df)
print(df.dtypes)
```

#### 3. Format parameter:
→ We can represent the data format in differcent ways ,

o March 23rd, 2015 as "03-23-15" or "3/23/2015" and etc.

→ So, we can use the format parameter to specify the exact format of the

string.


##### i. code:%y
- Description: full year
- Example : 2001

##### ii. code:%m
- Description: month with zero padding
- Example :04

##### iii. code:%d
- Description: day  with zero padding
- Example :07

##### iv. code:%I
- Description:Hour(12hr clock)/zero padding
- Example :02

##### v. code:%p
- Description:AM or PM
- Example :AM

##### vi. code:%M 
- Description: minutes with zero padding
- Example :05

##### vii. code:%s
- Description:Second with zero padding
- Example :09

#### 4) Nat values:

→ Date column may contains missing values in pandas Data frame.

→ If Data column contains missing values then while converting into Date data type then we will get an error.

→ By using errors = "coerce" keyword argument we can solve this problem.

→ This argument converts Date column missing Values into Nat (Not a Time) value.
→ Coerce еrrors i.e. convert un parcseable date into Nat (Not a Time)

```python
# creating dataframe
import pandas as pd

d = {
    "Product": ["Vivo", "Oppo", "Iphone"],
    "Status": ["success", "success", "Failed"],
    "cost": [10000, 12000, 13000],
    "Purchase_date": ["20190902", "20190916", "20190915"]
}

df = pd.DataFrame(d)

print(df)
print(df.dtypes)
```

```python
# convert Date with specific format:Error
import pandas as pd

d = {
    "Product": ["Vivo", "Oppo", "Iphone"],
    "Status": ["success", "success", "Failed"],
    "cost": [10000, 12000, 13000],
    "Purchase_date": ["2019-09-02", "2019-09-16", "2019-09-15"]
}

df = pd.DataFrame(d)

df["Purchase_date"] = pd.to_datetime(
    df["Purchase_date"], errors="coerce"
)

print(df)
print(df.dtypes)
```

#### 7) Extract year, month, day from Date column:

→ Based on requirement we can get year, month, day, hour, mintue from Date column.
 → Sometimes it can be useful to break up a column of dates into components .
 #### EG: Date value
 - I bought one product --> 02/14/2026 12:15:00
 - month --> 02
 - day -->14
 - year -->2026
 - hour -->12
 - min -->15
 - second -->00

```python
# breaking up the data column value into multiple features
import pandas as pd
df=pd.read_csv("veer.csv",parse_dates=["Purchase_date"])
df["year"]=df["Purchase_date"].dt.year
df["month"]=df["Purchase_date"].dt.month
df["day"]=df["Purchase_date"].dt.day
print(df)
```

#### 18. Encoding Days of the week:
- We can get the day of the week for each date by using pandas.

  ♡ Knowing the days names will be helpful to understand the bersiness / flow, like we can compare total sales on specific day for the past three year.



```python
#Encoding days of the week
import pandas as pd

data = {
    "status": ["success", "success", "failed"],
    "cost": [10, 20, 30],
    "purchase_date": ["2026-01-01", "2026-01-02", "2026-01-03"]
}

df = pd.DataFrame(data)

# Convert the column to datetime format
df["purchase_date"] = pd.to_datetime(df["purchase_date"])

print(df)

# Extract the day of the week
print(df["purchase_date"].dt.day_name())
```
