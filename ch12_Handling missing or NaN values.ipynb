#### 1.NaN value:
- The full form of NaN is Not a Number.
- The purpose of NaN is,to represent the missing values in data.
- The data type of NaN is float.
- While loading csv file,if file having missing values then it will be considered as NaN values.
- During data analysis we need to handle these NaN values.
  * For example, suppose different users surveyed may choose not to share their income,some user may choose not to share the address in the way datasets went missing.
#### None and NaN:
- __None__:- It is a python object which is holding nothing.
- __NaN__:-It is a pandas related object,which represents missing data.

#### 3.isna() and isnull() method-checking NaN values:
- isna() and isnull() are a predefined methods in dataframe.
- We can access these methods by using dataframe object.
- By using these methods we can check missing values exist in dataframe or not.
- If missing values are available then,it returns as True,otherwise false means boolean value.

```python
import pandas as pd

v= r"C:\Users\mehek\OneDrive\Desktop\DF File\students.csv"

df=pd.read_csv(v)

df1=df.isna()

print(df)

print(df1)

print(df1.head())
```

#### Note:
- isnull() and isna() both methods works in same way.
- We use isna() method in real time.
#### 4.notnull() method-checking NaN values:
- notnull() is a predifined method in dataframe.
- We can access this method by using dataframe object.
- By using this method we can check missing values exist in dataframe or not.
- If missing values are available then it returns as false,otherwise True.
- It is also not used in real time.

```python
import pandas as pd

v= r"C:\Users\mehek\OneDrive\Desktop\DF File\students.csv"

df=pd.read_csv(v)

df1=df.notnull()

print(df.head())

print(df1.head())
```

#### 5.counting NaN value in column wise:
- We can count number of missing values inn Dataframe.
- By using isna() and sum() methods we can count the number of missing values in each column.

```python
import pandas as pd

v= r"C:\Users\mehek\OneDrive\Desktop\DF File\students.csv"

df=pd.read_csv(v)

df1=df.notnull().sum()

print(df.head())

print(df1.head())
```

```python
# count the missing values in each columns with ppercentage
import pandas as pd

v= r"C:\Users\mehek\OneDrive\Desktop\DF File\students.csv"

df=pd.read_csv(v)

s=df.notnull().sum()

per=(s*100)/len(df)

print(per.head())
```

#### 6.dropna() method-handling missing value:
- dropna() is a predefined method in dataframe.
- We can access dropna() method by using dataframe object.
- This method drops the rows where atleast one value is missing.


```python
# dropping rows where NaN values existing
import pandas as pd

v= r"C:\Users\mehek\OneDrive\Desktop\DF File\students.csv"

df=pd.read_csv(v)

s=df.dropna()

print(s.head())
```

```python
# converting float columns type into integer type
import pandas as pd

v = r"C:\Users\mehek\OneDrive\Desktop\DF File\students.csv"

df = pd.read_csv(v)

s = df.dropna()

print("Data after removing missing values:")
print(s.head())

u = s.astype(int)

print("Data after converting numeric columns to integers:")
print(u.head())
```

#### 7.dropna(inplace=True) method handling missing values:
- dropna(inplace=True) is a predefined method in dataframe.
- we can access this method by using dataframe object.
- This method drops the rows and perform changes an existing dataframe.

```python
# droping NaN values by using inplace parameter
import pandas as pd
v = r"C:\Users\mehek\OneDrive\Desktop\DF File\students.csv"
df=pd.read_csv(v)
df.dropna(inplace=True)
print(df)
```

#### 8.fillna() method-handling missing values:
- It is a  predefined method in dataframe.
- We can access the method by using dataframe object.
- By using this method we can fill missing/NaN values with specific values.
  * Fillna(0)->This method fill NaN with "0".
  * Fillna(number)->This method fill NaN with number.

```python
import pandas as pd
v = r"C:\Users\mehek\OneDrive\Desktop\DF File\students.csv"
df=pd.read_csv(v)
df1=df.fillna(0)
print(df.head())
print(df1.head())
```
