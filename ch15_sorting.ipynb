#### (i)sort-values (by = "column name"):

→ sort-values (p) is a predefined method in dataframe class.

→ We should access this method by using Dataframe class.
→ This method sort the values based on column  which are we specified.

 * Number default sorting in ascending order.
 * string default sorting is alphabetical (dictionary) order.


```python
# sorting using sort_values(p) method in string
import pandas as pd
df=pd.read_csv("sales1.csv")
cond=df.sort_values(by="Name")
print(cond)
print(cond.head())
```

```python
# sorting using sort_values(p) method in numeric
import pandas as pd
df=pd.read_csv("sales1.csv")
cond=df.sort_values(by="Age")
print(cond)
print(cond.head())
```

```python
# sorting using sort_values(p) method in numeric descending order
import pandas as pd
df=pd.read_csv("sales1.csv")
cond=df.sort_values(by="Age",ascending=False)
print(cond)
print(cond.head())
```

##### here,
- by -->keyword argument
- ascending -->keyword argument

```python
# Loading file in company style
#(function with return statement)
import pandas as pd
def file(p):
    dataframe =pd.read_csv(p)
    return dataframe

df=file("sales1.csv")
print(df)
print(df.head())
```

#### 2.SORT_INDEX():

→ sort-index()) is prdefined method in Dataframe class.

```python
# if index we want to sort
import pandas as pd
df=pd.read_csv("sales1.csv")
cond=df.sort_index()
print(cond)
print(cond.head())
```
