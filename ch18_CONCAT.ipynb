```python
# creating a dataframe
import pandas as pd
d1=[[11,22],[44,55]]
d2=[[75,48],[65,68]]
c=["A","B"]
df1=pd.DataFrame(d1)
df2=pd.DataFrame(d2)
print(df1)
print(df2)
```

```python
# Creating DataFrames and using concat
import pandas as pd

d1 = [[11, 22], [44, 55]]
d2 = [[75, 48], [65, 68]]

c = ["A", "B"]

df1 = pd.DataFrame(d1, columns=c)
df2 = pd.DataFrame(d2, columns=c)

print(df1)
print(df2)

result = [df1, df2]
df3 = pd.concat(result)

print(df3)
```

```python
# if you want to ignore the previous index
import pandas as pd

d1 = [[11, 22], [44, 55]]
d2 = [[75, 48], [65, 68]]

c = ["A", "B"]

df1 = pd.DataFrame(d1, columns=c)
df2 = pd.DataFrame(d2, columns=c)

print(df1)
print(df2)

result = [df1, df2]
df3 = pd.concat(result,ignore_index=True)

print(df3)
```

#### here,
- ignore_index --> keyword argument
- concat -->function,because called vy aliasing

```python
# if you want to ignore the previous index
import pandas as pd

d1 = [[11, 22], [44, 55]]
d2 = [[75, 48], [65, 68],[55,77]]

c = ["A", "B"]

df1 = pd.DataFrame(d1, columns=c)
df2 = pd.DataFrame(d2, columns=c)

print(df1)
print(df2)

result = [df1, df2]
df3 = pd.concat(result,ignore_index=True)

print(df3)
```

```python
# if column names are not same
import pandas as pd

d1 = [[11, 22], [44, 55]]
d2 = [[75, 48], [65, 68],[55,77]]

c1 = ["A", "B"]
c2=["L","K"]
df1 = pd.DataFrame(d1, columns=c1)
df2 = pd.DataFrame(d2, columns=c2)

print(df1)
print(df2)

result = [df1, df2]
df3 = pd.concat(result,ignore_index=True)

print(df3)
```

```python
# if column names are not same you don't want NaN values
import pandas as pd

d1 = [[11, 22], [44, 55]]
d2 = [[75, 48],[55,77]]

c1 = ["A", "B"]
c2=["L","K"]
df1 = pd.DataFrame(d1, columns=c1)
df2 = pd.DataFrame(d2, columns=c2)

print(df1)
print(df2)

result = [df1, df2]
df3 = pd.concat(result,axis=1)

print(df3)
```
