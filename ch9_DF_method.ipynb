#### 2. Accessing single columns from dataframe:
- We can access column from data frame;
  * If we access single column then it returns the series.
  * If we access two column then it returns the dataframe with two column.

```python
# single column
import pandas as pd
file=r"C:\Users\mehek\OneDrive\Desktop\DF File\sales1.csv"
df=pd.read_csv(file)
print("Display the column")
print(df)
total=df["Name"]
print(total)
```

```python
# Accessing two columns
import pandas as pd

file = r"C:\Users\mehek\OneDrive\Desktop\DF File\sales1.csv"
df = pd.read_csv(file)

print("Display the result")
print(df)

print(df[["Name", "Age"]])
```

```python
# Converting Dataframe to CSV:-
Syntax: df.to_csv("result.csv")
#EX:
import pandas as pd
file=r"C:\Users\mehek\OneDrive\Desktop\DF File\sales1.csv"
df=pd.read_csv(file)
cols=["Name","Age"]
print(df[cols])
df.to_csv("result.csv")
```

```python
# Accessing single column from the dataframe ,apply sum:
import pandas as pd 
df=pd.read_csv("result.csv")
total=df["Age"]
print(total)
print(total.sum())
```
