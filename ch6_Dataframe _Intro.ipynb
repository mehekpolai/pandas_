#### a. Creating Empty DataFrame:
- We can create an empty dataframe.

```python
import pandas as pd
df=pd.DataFrame()
print(df)
print(type(df))
```

#### b. By using single list:
- We can create dataFrame by using singlelist.
- If we are using single list then its single column dataframe.
- If we are using list of lists (nested list) then it's multiple column if there is no name is provide then by default column name is 0.
  

```python
# using single list
import pandas as pd
b=[44,55,88,77]
df=pd.DataFrame(b)
print(df)
print(type(df))
```

#### Note:
- From the o/p ,Dataframe created single column.
- Here,column name is "Zero" ,we can customise this as well.
#### Note on index:
- If no index is passed,then by default,index will be Range(n), where n is the array length.

```python
import pandas as pd
b=["hari","rama","paro","gita"]
df=pd.DataFrame(b)
print(df)
print(type(df))
```

```python
# checking length:
import pandas as pd
b=["hari","rama","paro","gita"]
df=pd.DataFrame(b)
print(df)
print("The length is:",len(df))
```

#### c. By using nested list
- we can create dataframe with list of lists(nested list).
- If we are using list of lists then it create a dataframe with multiple columns.

```python
import pandas as pd
b=[["hari",45],["rama",44],["paro",41],["gita",39]]
df=pd.DataFrame(b)
print(df)
```

#### note:
- From the o/p dataframe created with two columns.
- here, columns are 0 and 1 and we can customize this as well.
#### Giving column names to Dataframe:
- We can give column names to dataframe.
  

```python
import pandas as pd
b=[["hari",45,12000],
   ["rama",44,41500],
   ["paro",41,41000],
   ["gita",39,42100]]
c=["name","age","salary"]
df=pd.DataFrame(b,columns=c)
print(df)
```

#### here,
- details ---> nested list
- c ---->list
- df --->dataframe object
- column ---? keyword argument
#### d. By using Dictionary:
- WE can create dataframe by using dictionary.
- IF we using list of lists then it create a dataframe with multiple columns.

```python
import pandas as pd
b={"name":["k","m","j","f","r"],
  "age":[12,45,74,10,65]}

df=pd.DataFrame(b)
print(df)
```

- In the above example Name and age considered as columns names.those are keys of dictionary.
#### We can customise the index values:
- By default index value starts with 0.
- We can customize the index values in dataframe.
- If index is passed ,then the length of index should equal  to the llength of the dataframe.

```python
# byy using nested list indexing
import pandas as pd
b=[["hari",45,12000],
   ["rama",44,41500],
   ["paro",41,41000],
   ["gita",39,42100]]
c=["name","age","salary"]
i=[11,44,55,22]
df=pd.DataFrame(b,columns=c,index=i)
print(df)
```

```python
import pandas as pd
b=[["hari",45,12000],
   ["rama",44,41500],
   ["paro",41,41000],
   ["gita",39,42100]]
c=["name","age","salary"]
i=["Row1","Row2","Row3","Row4"]
df=pd.DataFrame(b,columns=c,index=i)
print(df)
```
