#### 1.Groupby Introduction:

→ Groupby is very common and important operations in Data analysis.

→ There are mainly 3 steps in Group by operations, 
  o Splitting the data into groups based on some criteria.

  o Applying operation to each group independently.

  o Combaining the results.

  
→ for Example,

  o If we apply, groupby on productNam then it groups the data into based on name of the product.

→ The groupby (p) method returns a GroupBy object. 

#### 2.groupby(p):

+ groupBy(p) is predefined method, in Dataframe.

+ We should access this method by using Dataframe object.

+ This method return's GroupBy object.

```python
# creating a product dataframe 
import pandas as pd
d={
    "product name":["Samsung","Nokia","vivo","oppo","Samsung","Samsung","vivo","Iphone"],
    "order":[12,41,42,15,10,41,12,11]
}
df=pd.DataFrame(d)
print(df)
```

```python
# apply groupby method in product name wise

import pandas as pd
d={
    "product_name":["Samsung","Nokia","vivo","oppo","Samsung","Samsung","vivo","Iphone"],
    "order":[12,41,42,15,10,41,12,11]
}
df=pd.DataFrame(d)
print(df)
a=df.groupby(["product_name"])  # spilt the data
print(a)
print(a.sum())  # apply operation
```

```python
# apply groupby method in order wise

import pandas as pd
d={
    "product_name":["Samsung","Nokia","vivo","oppo","Samsung","Samsung","vivo","Iphone"],
    "order":[12,41,42,15,10,41,12,11]
}
df=pd.DataFrame(d)
print(df)
a=df.groupby(["order"])  # spilt the data
print(a)
print(a.sum())  # apply operation
```

```python
# apply single function to column name
# step 1 : importing library
import pandas as pd
# step 2 :loadinf dataframe
df=pd.read_csv("product.csv")
# step 3 :create a dictionary :- We can directly put sum() method in any numeric value 
d={"product_cost":sum}
# step 4 :creating list of string
cols=["date","product_name"]
# step 5 :apply groupby method
a=df.groupby(cols)
# step 6 :apply aggregate operation
b=a.agg(d)
print(b)
```

###### Here,
  * sum -> function,coming from python
  * agg -> method, coming from pandas
   

```python
# apply single function to column name

import pandas as pd

df=pd.read_csv("product.csv")
 
d={"product_cost":sum,"product_name":"count"}

cols=["date","product_name"]

a=df.groupby(cols)

b=a.agg(d)
print(b)
```

```python
# apply single function to column name

import pandas as pd

df=pd.read_csv("product.csv")
 
d={"product_cost":[min,max,sum]}

cols=["product_cost"]

a=df.groupby(cols)

b=a.agg(d)
print(b)
```
