#### 1.NaN:
- The full form of NaN is Not a Number.
- The purpose of NaN is ,to represent the missing values in data.
- The data type of NaN is __float__.
- We will learn more about NaN in 12th chapter.
#### 2.None,NaN,nan and null:
- None,NaN,nan and null are synonyms.
- These all are representing to empty or missing data found in service,dataframe .

```python
import pandas as pd
import numpy as np
marks=[86,61,43,76,19]
s=pd.Series(marks)
print(s)
```

```python
import pandas as pd

marks=[11,42,98,76,48,75]
s=pd.Series(marks)
print(s)
```

```python
# creating series with NaN values
import pandas as pd
import numpy as np
marks=["dezzy","leeisa",None,"ariya",np.nan]
s=pd.Series(marks)
print(s)
```
