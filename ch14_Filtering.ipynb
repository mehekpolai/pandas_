#### 2.By using relational operator
→ By using relational operators we can apply the filter on dataframe.


```python
# using relational operator single condition
import pandas as pd
df=pd.read_csv("nirma.csv")
cond=df["salary"]<21000
print(df[cond])
```

```python
# using relational operator multiple condition
import pandas as pd
df=pd.read_csv("nirma.csv")
cond1=df["salary"]<21000
cond2=df["salary"]>16000
print(df[cond1 & cond2])
```

```python
# using relational operator multiple condition
import pandas as pd

df = pd.read_csv("nirma.csv")

cond1 = df["name"] == "mahi"
cond2 = df["salary"] == 19650

print(df[cond1 & cond2])
```

#### 3. Filtering by thas using loc and iloc:

→ We can filters the dataframe by using loc and iloc indexers as well.


```python
#  multiple condition using loc
import pandas as pd

df = pd.read_csv("nirma.csv")

cond1 = df["name"] == "mahi"
cond2 = df["salary"] == 19650

print(df.loc[cond1 & cond2])
```

```python
# using iloc 1st 5 rows and all column
import pandas as pd

df = pd.read_csv("nirma.csv")

cond = df.iloc[0:5, ]

print(cond)
```

#### (ii) Selecting multiple valus of a column:

→ We can filtere dataframe by, providing multiple values of a column.

```python
# access multiple ccondition using and("&") operator and loc
import pandas as pd

df = pd.read_csv("nirma.csv")

cond1 = df["name"] == "mahi"
cond2 = df["salary"] == 19650

print(df.loc[cond1 & cond2])
```

```python
# access multiple ccondition using or("|") operator and loc
import pandas as pd

df = pd.read_csv("nirma.csv")

cond1 = df["name"] == "mahi"
cond2 = df["salary"] == 19650

print(df.loc[cond1 | cond2])
```

#### (iii) isin() method:

→ It is predefined method in Series class.

→ We should access this method by using Series object.

→ By using this method we can select date from DataFrame.

```python
# using loc isin() method
import pandas as pd

df = pd.read_csv("nirma.csv")

cond1 = ["mahi"]
cond2 = df["name"].isin(cond1)

print(df.loc[cond2])
```

```python
# using loc isin() method access 2 values
import pandas as pd

df = pd.read_csv("nirma.csv")

cond1 = ["mahi","rishab"]
cond2 = df["name"].isin(cond1)

print(df.loc[cond2])
```

#### 4.Select Non-missing Data from Dataframe notnull() method:

→ It is predefined method in series class.

→ We should access this method by using Series object.

→ By using this function we can select the Dataframe which having non NaN values.

```python
import pandas as pd

df = pd.read_csv("nirma.csv")

cond1 =df["age"].notnull()

print(cond1)
print(len(cond1))
```
