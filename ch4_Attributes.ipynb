### A.Series Attributes:
- Series is a pre-defined class.
- Series having different attributes.
- These attributes return information about the object.

1. values:
- values is represents attributes in series.
- We can access valuesw attributes by using series object.
- This attribute returns the group of values an array.

```python
import pandas as pd
a=[10,20,30,45,75,146]
s=pd.Series(a)
print("Series object is created and loaded with data,object name is,s")
print(s)
print(s.values)
```

#### here,
- s ---> Series object
- values ---> attribute
- import ---> keyword
- print ---> keyword
- a ---> variable
- series --->class

```python
# accessing values attributes
a=["vibya","jiya","Sankar","jit"]
s=pd.Series(a)
print("Series object is created and loaded with data,object name is,s")
print(s)
print(s.values)
```

2. index:
- index is pre-defined attribute in Series class.
- We can access index attributes by using series object.
- This attribute returns the index range like, RangeIndex(start=0,stop=6,step=1).
* Q.Is value and index attributes are also called as instance variable?proove it?

```python
import pandas as pd
a=["vibya","jiya","Sankar","jit"]

s=pd.Series(a)
print(s)
print(s.values)
print(s.index)
```

```python
import pandas as pd
a=["vibya","jiya","Sankar","jit"]
m=[1100,52,84,22]
s=pd.Series(a,index=m)
print(s)
print(s.values)
print(s.index)
```

```python
import pandas as pd
marks=[115,775,441,881]
a=["vibya","jiya","Sankar","jit"]

s=pd.Series(marks,index=a)
print(s)
print(s.values)
print(s.index)
```

3. dtypes:
- dtyes is pre-defined attribute in Series class.
- We can access index attributes by using series object.
- This attribute returns the data type of series column.

```python
import pandas as pd
marks=[115,775,441,881]
s=pd.Series(marks)
print(s)

print(s.dtype)
```

```python
import pandas as pd
marks=[115,775,441,881]
s=pd.Series(marks)
print("I have created servies object name is :s")
print(s)
print("How to get the data:")
print(s.values)
print("how to get the index:")
print(s.index)
print("how to get the data type:")
print(s.dtype)
```

4. size:
- size is pre-defined attribute in Series class.
- We can access index attributes by using series object.
- This attribute returns the number of values in series.

```python
import pandas as pd
marks=[115,775,441,881]
s=pd.Series(marks)
print("how many values in  a object:")
print(s)
print(len(s)) # function from python
print(s.size) # attribute from pandas
```
