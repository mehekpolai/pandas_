#### i.CSV file:
- CSV stands for comma separated file.
- A CSV file contains data like rows and columns.
- Whenever we are loading csv files then pandas returns dataframe.
- We can load csv file by using read_csv("file name with path")function.
- here, read_csv = function name

```python
import pandas as pd
df=pd.read_csv("sales1.csv")
print(df)
```

#### iii. TSV file
- It stands for tab separated file.
- Whenever we are loading tsv file  then pandas returns dataframe.
- We can load csv by using read_table("filename with path") function.
- Syntax: pd.read_table(path and filename)
### How can we make TSV:
- open notebook.
- Write there like:-(separated by tab)
-  __Name__  __location__  __age__
-  j       jammu    12
-  k       kerala   13
-  l       lakhno   10
-  save as :example.tvs

```python
import pandas as pd
file=r"C:\Users\mehek\OneDrive\Desktop\DF File\sales1.tsv"
df=pd.read_table(file)
print(df)
```
