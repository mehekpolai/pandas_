#### (i) innere Merge / Inner join:

→ We can perform inner join on Dataframe.

→ This inner join is a kind of intersection means, it keeps the common data.

#### Merge (df1, df2, on="column", how="type")

→ merge (df1, df2, on="column", how="type") is predefined function in pandas.

→To perform inner join, we need to pass how values as "inner" as per the syntax.

#### Syntax:
- pd.merge (df1, df2, on="column", how="inner")

```python
# Creating two DataFrames using dictionaries
import pandas as pd

d1 = {
    "id": [11, 12, 15, 14, 17],
    "Name": ["veeru", "charu", "bina", "panni", "lily"],
    "sub": ["math", "hindi", "sanskrit", "geo", "phy"]
}

d2 = {
    "id": [447, 448, 556, 114, 775],
    "Name": ["vidya", "cherry", "binay", "patnika", "lichu"],
    "sub": ["math", "hindi", "sanskrit", "chem", "phy"]
}

df1 = pd.DataFrame(d1)
df2 = pd.DataFrame(d2)

print(df1)
print(df2)
```

```python
# inner join
import pandas as pd

d1 = {
    "id": [11, 12, 15, 14, 17],
    "Name": ["veeru", "charu", "bina", "panni", "lily"],
    "sub": ["math", "hindi", "sanskrit", "geo", "phy"]
}

d2 = {
    "id": [447, 448, 556, 114, 775],
    "Name": ["vidya", "cherry", "binay", "patnika", "lichu"],
    "sub": ["math", "hindi", "sanskrit", "chem", "phy"]
}

df1 = pd.DataFrame(d1)
df2 = pd.DataFrame(d2)

print(df1)
print(df2)

inner_join=pd.merge(df1,df2,on="sub",how="inner")

print(inner_join)
```

#### here,
- __on__ --> keyword argument,used for based on which column you want join
- __how__--> keyword argument,used for which type of join
#### (ii) Left join:
→ keep every row in the left dataframe.
→ Where there are missing value of the "on" variable in the right dataframe Filled with NaN value is the result.

```python
# left join
import pandas as pd

d1 = {
    "id": [11, 12, 15, 14, 17],
    "Name": ["veeru", "charu", "bina", "panni", "lily"],
    "sub": ["math", "hindi", "sanskrit", "geo", "phy"]
}

d2 = {
    "id": [447, 448, 556, 114, 775],
    "Name": ["vidya", "cherry", "binay", "patnika", "lichu"],
    "sub": ["math", "hindi", "sanskrit", "chem", "phy"]
}

df1 = pd.DataFrame(d1)
df2 = pd.DataFrame(d2)

print(df1)
print(df2)

left_join=pd.merge(df1,df2,on="sub",how="left")

print(left_join)
```

#### id-y values are float, because Nan Value is present there.

#### (iii) Right join

→ keep every row in the right dataframe.
→ Where there, are missing values of the "on" variable in the Left column filled with NaN Values in the result.

```python
# Right join
import pandas as pd

d1 = {
    "id": [11, 12, 15, 14, 17],
    "Name": ["veeru", "charu", "bina", "panni", "lily"],
    "sub": ["math", "hindi", "sanskrit", "geo", "phy"]
}

d2 = {
    "id": [447, 448, 556, 114, 775],
    "Name": ["vidya", "cherry", "binay", "patnika", "lichu"],
    "sub": ["math", "hindi", "sanskrit", "chem", "phy"]
}

df1 = pd.DataFrame(d1)
df2 = pd.DataFrame(d2)

print(df1)
print(df2)

right_join=pd.merge(df1,df2,on="sub",how="right")

print(right_join)
```

#### (iv) full outer join

→ A full outer join returns all the rows from the left and right Dataframe.

→ Where there is no common data there it will be fills with NaN values .

```python
# full outer join
import pandas as pd

d1 = {
    "id": [11, 12, 15, 14, 17],
    "Name": ["veeru", "charu", "bina", "panni", "lily"],
    "sub": ["math", "hindi", "sanskrit", "geo", "phy"]
}

d2 = {
    "id": [447, 448, 556, 114, 775],
    "Name": ["vidya", "cherry", "binay", "patnika", "lichu"],
    "sub": ["math", "hindi", "sanskrit", "chem", "phy"]
}

df1 = pd.DataFrame(d1)
df2 = pd.DataFrame(d2)

print(df1)
print(df2)

full_outer_join=pd.merge(df1,df2,on="sub",how="outer")

print(full_outer_join)
```

#### 4.other types of join

(i) one to one

(ii) Many to one

(iii) many to many

(i) one to one:

→ This is very simple join and similar to the column-wise concatenation.

(ii) Many to one:

→ Many-to-one joins  are joins in which one of the two key columns contains duplicate entries.

(iii) many to many:

→If the key column in both the left and right Dataframe contains duplicates, then the result is a many merge.
