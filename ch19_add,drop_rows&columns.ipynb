#### 1. Adding column to Dataframe:
- Based on requirement we can add column to existing dataframe.

```python
# loading the dataframe
import pandas as pd
df=pd.read_csv("student.csv")
print(df)
```

```python
# loading the dataframe
import pandas as pd
df=pd.read_csv("student.csv")
print(df)
```

```python
# Add new column "Total"
import pandas as pd
df=pd.read_csv("student.csv")
df["Total"] = df["Product_cost"] * df["Quantity"]
print(df)
```

```python
# Add new column "Total" by using apply method
import pandas as pd
df=pd.read_csv("student.csv")
def Total(df):
    t=  df["Product_cost"] * df["Quantity"]
    return t
df["Total"] = df.apply(Total,axis=1)
print(df)
```

```python
# Add new column "Total" in specific position
import pandas as pd

df = pd.read_csv("student.csv")

# Calculate Total
new = df["Product_cost"] * df["Quantity"]

# Insert Total at position 2
df.insert(2, "Total", new) #df.insert("position","column name","operation")

print(df)
```

#### 2. Dropping column from Dataframe:
- Based on requirement we can drop column from existing dataframe.

```python
 # drop single column
import pandas as pd
df=pd.read_csv("student.csv")
df1=df.drop(columns="Quantity")
print(df)
print(df1)
```

```python
# drop multiple columns
import pandas as pd
df=pd.read_csv("student.csv")
c=["Product_cost","Quantity"]
df1=df.drop(columns=c)
print(df.head())
print(df1.head())
```

```python
# drop multiple columns
import pandas as pd
df=pd.read_csv("student.csv")
c=["Product_cost","Quantity","Product_Name"]
df1=df.drop(columns=c)
print(df.head())
print(df1.head())
```

#### 3. Dropping rows from dataframe:
- We can drop the rows from dataframe by using drop method.

```python
# drop single row
import pandas as pd
df=pd.read_csv("student.csv")
print(df.head())
df1=df.drop(3,axis=0)
print(df1.head())
```

```python
# drop multiple row
import pandas as pd
df=pd.read_csv("student.csv")
print(df)
r=[3,6,7,8]
df1=df.drop(r,axis=0)
print(df1)
```
