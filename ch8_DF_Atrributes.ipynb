#### 10 T(constant):
- It is predifined attributes in dataframe class.
- we can access T attribute by using dataframe object.
- T means its transpose, it returns rows as column and columns as rows.

```python
import pandas as pd
details=[["damu",1000],
        ["hari",2000],
        ["jenila",3000],
        ["herry",4000]]
df=pd.DataFrame(details)
print(df)
print(df.T)
```
