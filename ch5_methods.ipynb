####  Series methods:
- Series is a  predifined class.
- Series class having different methods.
- These methods performs operations on Series of values.

1. head()
- It is predefined method in Series class.
- We can access head() method by using series object.
- This method returns first five values from series.

```python
# acessing 1st 5 values from series
import pandas as pd
m=[10,20,30,40,50,60,70,80,90]
s=pd.Series(m)
print("How to get 1st 5 values")
print(s)
print(s.head())
print(s.head(7))
# if we want to customise our own values then we input it beacause by default value is 5 values.
```

#### 2. tail()
- It is predefined method in Series class.
- We can access tail() method by using series object.
- This method returns last five values from series.

```python
# acess last 5 values
import pandas as pd
m=[10,20,30,40,50,60,70,80,90]
s=pd.Series(m)
print("How to get last 5 values")
print(s)
print(s.tail())
print(s.tail(7))
```

#### 3. sum()
- It is predefined method in Series class.
- We can access sum() method by using series object.
- This method returns the sum of all values from series.

```python
import pandas as pd
m=[10,20,30,40,50,60,70,80,90]
s=pd.Series(m)
print("sum of all values")
print(s)
print(s.sum())
```

```python
# sum using NaN values
import pandas as pd
m=[10,20,None,40,50,None,70,80,90]
s=pd.Series(m)
print("find how many customers")
print(s)
print(s.size)
print("find how many null values present")
print(s.isnull().sum())
print("sum of all  customers")
print(s.sum())
```

#### 4. count()
- It is predefined method in Series class.
- We can access count() method by using series object.
- This method returns the number of non/Nan/null values/ values from series.

```python
import pandas as pd
m=[101,120,156,750,123,452]
s=pd.Series(m)
print("information")
print(s)
print("how many days")
print(s.count())
```

```python
import pandas as pd
m=["nilu","lilu","pilku","gilu"]
s=pd.Series(m)
print("information")
print(s)
print("how many days")
print(s.count())
```

#### 5. mean()
- It is predefined method in Series class.
- We can access mean() method by using series object.
- This method returns the mean of series of values.

```python
import pandas as pd
m=[44,45,75,48,42]
s=pd.Series(m)
print(s)
print("average of m")
print(s.mean())
```

#### 6. describe()
- It is predefined method in Series class.
- We can access descibe() method by using series object.
- This method the below values:-
  * count
  * mean
  * std
  * min
  * 25%
  * 50%
  * 75%
  * max

```python
import pandas as pd
q=[11,44,77,88,55,22,33,66]
s=pd.Series(q)
print(s)
print(s.describe())
```

#### 7. unique()
- It is predefined method in Series class.
- We can access unique() method by using series object.
- This method returns the values from the series. 

```python
import pandas as pd
p=[11,44,77,11,44,88,55]
s=pd.Series(p)
print(s)
print(s.unique())
```

#### 8. nunique()
- It is predefined method in Series class.
- We can access nunique() method by using series object.
- This method returns the number of unique values from the series. 

```python
import pandas as pd
p=[11,44,77,11,44,88,55]
s=pd.Series(p)
print(s)
print(s.nunique())
```
