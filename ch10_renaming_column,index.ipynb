#### 1.renaming(p) method-changing single column name:
- rename(p) is a predifined method in dataframe.
- We can access rename() method by using dataframe object.
- By using this method we can  change single column in dataframe.

```python
# input :add the folder name where sales3 .csv file present
import pandas as pd
file=r"DF files\sales3.csv"
df1=pd.read_csv(file)
d={"ord id":"order_id"}
print("Before change:")
df1.head()# use head()method because i want to see 5 values
df2=df1.rename(columns=d)
print("After change:")
df2.head()# we can write df2.head() instead of print(df2.head)
```

#### 2.renaming(p) method-changing multiple column names:
- rename(p) is  a predefined method in dataframe.
- We can access rename() method by using dataframe object.
- We can even change multiple column names by using rename(p) method.
- We need to use python dictionary to specify old column as key and new column as value.
- It uses dictionary data type.

```python
import pandas as pd
file=r"C:\Users\mehek\OneDrive\Desktop\DF File\sales3.csv"
df1=pd.read_csv(file)
d={"ord id":"order_id",
  "cust name":"customer_name",
  "prod name":"product_name"
  }
print("Before change:")
print(df1.head())
df2=df1.rename(columns=d)
print("After change:")
print(df2.head())
```

#### 3.column attribute- changing multiple column name:
- columns is predefined attribute in dataframe class.
- We can access columns attribute by using dataframe object.
- By using columns attribute we can even changing multiple column names.

```python
file = r"DF files\sales3.csv"
df = pd.read_csv(file)

df.columns = [
    "Order_id",
    "Customer_name",
    "Product_name",
    "Product_cost",
    "Quantity"
]

print(df.head())
```

#### NOTE:
- While using columns attributes to change columns names then number off dataframe column names should be match with existing columns otherwise we will get error.
- In this trick we have to access all the column name otherwise we will get error.
#### 4.rename(p) method changing index:
+ rename(p) is predefined method in dataframe.
+ We can access rename() method by using dataframe object.
+ Based on requirement we can change the  index in dataframe.
+ By using rename(p) we can change the dataframe index.
+ It is also used dictionary.

```python
import pandas as pd
d={"Name":["Raj","taj","hari","natri"],
  "gender":["m","m","m","f"],
      "age":[12,15,42,10]}
i={0:14,1:42,2:77,3:70}
df1=pd.DataFrame(d)
df2=df1.rename(index=i)
print(df1)
print(df2)
```

#### 5.index attribute-changing multiple column name:
- index is predefined attribute in dataframe class.
- We can access index attribute by using dataframe object..
- By using index attribute we can even change index in dataframe.


```python
import pandas as pd
file=r"DF files\sales3.csv"
df1=pd.read_csv(file)
df2=df1.head()
df2.index=[11,42,15,75,48]
print(df2)
```

#### 6.converting column name lower case to upper case:
- Based on requirement we can convert dataframe column names from lower case to upper case.

```python
import pandas as pd
file=r"DF files\sales3.csv"
df1=pd.read_csv(file)

df1.columns=df1.columns.str.upper()
print(df1)
```
